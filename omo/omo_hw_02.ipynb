#### Задание 1. Соберите всю информацию о студентах в одну таблицу df. В получившейся таблице должна быть информация и оценки всех студентов из всех групп. Напечатайте несколько строк таблицы для демонстрации результата.¶

```python
import requests, zipfile, io
df1 = pd.DataFrame()
df2 = pd.DataFrame()

a = requests.get('https://github.com/hse-ds/iad-intro-ds/raw/master/2023/homeworks/Data.zip')

with open ('Data.zip', 'wb') as b:
    b.write(a.content)

with zipfile.ZipFile('Data.zip') as f:
    f.extractall()

for i in range(10):
    with open (f'Data\Students_info_{i}.csv') as table:
        df1 = pd.concat([df1, pd.read_csv(table)], ignore_index = True)

for i in range(10):
    with open (f'Data\Students_marks_{i}.csv') as table1:
        df2 = pd.concat([df2, pd.read_csv(table1)], ignore_index = True)
df = pd.merge(df1, df2, on = 'index')
df.head(7)
```

**Задание 10. Ответьте на вопросы:**

```python
#  Какая доля студентов сдала экзамен по математике?

sdali = np.sum(df['math score'] >= passmark)
dolya = sdali / df.shape[0]
print(dolya, '- доля сдавших экзамен по математике.')

# Какая доля студентов, проходивших курс подготовки к экзамену, сдала экзамен по математике?

completed_test = df.loc[df['test preparation'] == 'completed']
dolya2 = np.sum(completed_test['math score'] >= 50) / completed_test.shape[0]
print(dolya2, '- доля сдавших экзамен по математике, при этом проходивших курс подготовки.')

# Какая доля женщин, не проходивших курс подготовки к экзамену, не сдала экзамен по математике?

female = df.loc[df['gender'] == 'female']
female_none = female.loc[female['test preparation'] == 'none']
dolya3 = np.sum(female_none['math score'] < 50) / female_none.shape[0]
print(dolya3, '- доля женщин, которые не проходили курс и не сдали экзамен')
```

**Задание 12. Выполните задание 11 с помощью циклов. Сравните время выполнения.**

```python
%%time
# Для каждой этнической группы выведите средний балл за экзамен по чтению
summ = 0
count = 0

race = df['race/ethnicity'].unique()
for value in race:
    for i in range(df.shape[0]):
        if value == df['race/ethnicity'][i]:
            summ += df['reading score'][i]
            count += 1
    print('Среднее значение в', value, '=', summ / count)
    summ = 0
    count = 0
```

```python
%%time
# Для каждого уровня образования выведите минимальный балл за экзамен по письму
import math
minv = math.inf

edu = df['education'].unique()
for value1 in edu:
    for j in range(df.shape[0]):
        if value1 == df['education'][j]:
            if df['writing score'][j] < minv:
                minv = df['writing score'][j]
    print(f'Для уровня образования {value1} минимальный балл за экзамен по письму = {minv}')
    minv = math.inf
```

#### Задание 14. Сколько студентов успешно сдали экзамен по математике?

Создайте новый столбец в таблице df под названием Math_PassStatus и запишите в него F, если студент не сдал экзамен по математике (балл за экзамен < passmark), и P иначе.

Посчитайте количество студентов, сдавших и не сдавших экзамен по математике.

Сделайте аналогичные шаги для экзаменов по чтению и письму.

```python
def Passed (mark, subject):
    if mark[subject] >= passmark:
        return 'P'
    else:
        return 'F'
# Математика
df['Math_PassStatus'] = df.apply(lambda x: Passed(x, 'math score'), axis = 1)
print('Количество студентов, которые сдали экзамен по математике = ', np.sum(df['Math_PassStatus'] == 'P'))
print('Не сдали экзамен по математике: ', np.sum(df['Math_PassStatus'] == 'F'))
```

```python
# Чтение
df['Reading_PassStatus'] = df.apply(lambda x: Passed(x, 'reading score'), axis = 1)
print('Количество студентов, которые сдали экзамен по чтению = ', np.sum(df['Reading_PassStatus'] == 'P'))
print('Не сдали экзамен по чтению: ', np.sum(df['Reading_PassStatus'] == 'F'))
```

```python
# Письмо
df['Writing_PassStatus'] = df.apply(lambda x: Passed(x, 'writing score'), axis = 1)
print('Количество студентов, которые сдали экзамен по письму = ', np.sum(df['Writing_PassStatus'] == 'P'))
print('Не сдали экзамен по письму: ', np.sum(df['Writing_PassStatus'] == 'F'))

df.head()
```

#### Задание 15. Сколько студентов успешно сдали все экзамены?

Создайте столбец OverAll_PassStatus и запишите в него для каждого студента 'F', если студент не сдал хотя бы один из трех экзаменов, а иначе 'P'.

Посчитайте количество студентов, которые сдали все экзамены.

```python
def overall(marks):
    if marks['Math_PassStatus'] == 'P' and marks['Reading_PassStatus'] == 'P' and marks['Writing_PassStatus'] == 'P':
        return 'P'
    else:
        return 'F'

df['OverAll_PassStatus'] = df.apply(overall, axis = 1)
print('Количество студентов, которые сдали все экзамены = ', np.sum(df['OverAll_PassStatus'] == 'P'))
df.head()
```

#### Задание 16. Переведем баллы в оценки

### Система перевода баллов в оценки
####    больше 90 = A
####      80-90 = B
####      70-80 = C
####      60-70 = D
####      50-60 = E
####    меньше 50 = F (Fail)

Создайте вспомогательную функцию, которая будет по среднему баллу за три экзамена выставлять оценку студенту по данным выше критериям.

Создайте столбец Grade и запишите в него оценку каждого студента.

Выведите количество студентов, получивших каждую из оценок.

**В случае, если средний балл попадает на границу между оценками (т.е. равен ровно 60, 70 или 80 баллов), вы можете интерпретировать условие на своё усмотрение (т.е. можете поставить за 60 баллов оценку D, а можете - E).**

```python
def GetGrade(average_mark):
    if average_mark > 90:
        return 'A'
    elif average_mark > 80 and average_mark <= 90:
        return 'B'
    elif average_mark > 70 and average_mark <= 80:
        return 'C'
    elif average_mark > 60 and average_mark <= 70:
        return 'D'
    elif average_mark >= 50 and average_mark <= 60:
        return 'E'
    else:
        return 'F'

df['Grade'] = df[['math score', 'reading score', 'writing score']].mean(axis = 1).apply(GetGrade)

df['Grade'].value_counts() # количество студентов, получивших каждую из оценок.
```
